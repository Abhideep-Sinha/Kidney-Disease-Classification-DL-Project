```python
#ensure annotations
def get_product(x:int,y:int)->int:
    return x*y
```

```python
get_product(2,3)
```

```python
get_product(2,'3') #it is a bug because we are passing a string instead of int. but python will not give any error because it is dynamically typed language. so to avoid this we can use type annotations.
```

```python
from ensure import ensure_annotations

@ensure_annotations
def get_product(x:int,y:int)->int:
    return x*y
```

```python
get_product(2,3)
```
